# Hill Climbing algorithm 
is a local search optimization technique used in Artificial Intelligence to find a peak (maximum) or valley (minimum) in a target mathematical function. It evaluates neighboring points and moves iteratively in the direction of the highest improvement.

```python
def heuristic(x):
    return -(x - 6) ** 2 + 24

def neighbourStates(x):
    return [x-1, x+1]

def values(x):
    return heuristic(x)

def hillClimbing(start):
    current = start
    while True:

        nextStates = neighbourStates(current)
        neighbour = max(nextStates, key=heuristic)

        if values(neighbour) <= values(current):
            return current
        
        current = neighbour
    
result = hillClimbing(0)

print("Best state: ", result)
print("Best value: ", heuristic(result))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 12, 5000)
y = -(x - 6) ** 2 + 24

plt.plot(x, y)
plt.xlabel("state")
plt.ylabel("State Value")
plt.grid(True)

plt.scatter(result, heuristic(result))

plt.axvline(result, linestyle="--")
plt.axhline(heuristic(result), linestyle="--")


plt.show()
```
